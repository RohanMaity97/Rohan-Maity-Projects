```python
#Importing the necessary libraries and modules for Analysis and Visualization

import pandas as pd
import numpy as np
import plotly.express as px
import matplotlib.pyplot as plt
from plotly.subplots import make_subplots
import plotly.graph_objects as go
%matplotlib inline

from plotly.offline import init_notebook_mode
init_notebook_mode(connected=True)
pd.set_option('display.max_columns', None)
```

```python
#Loading the dataset and checking it

df = pd.read_csv("footwear_sales_data.csv")

df.head()
```

```python
#Checking the dataset's general information

df.info()
```

```python
#Since some issues have been identified with the dataset, let's fix it. 
#First and foremost, stripping the column names of any whitespace so that it wont cause any problems

df.columns = df.columns.str.strip()

#Converting the dates to datetime format

df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.strftime('%b')
df['Quarter'] = 'Q' + df['Order Date'].dt.quarter.astype(str)
df['Day of Week'] = df['Order Date'].dt.day_name()

#Converting the relevant columns from strings to float

columns_conversion = ['MRP (INR)', 'Sold Price (INR)', 'Revenue (INR)', 'Unit COGS (INR)', 'Net Revenue (INR)', 'Total COGS (INR)', 'Gross Profit (INR)']
for col in columns_conversion:
    df[col] = df[col].astype(str).str.replace(',', '').astype(float)
    
#Converting the Discount % column from string to integer
    
df['Discount (%)'] = (
    pd.to_numeric(
        df['Discount (%)'].str.replace('%', '', regex=False)
    ) / 100
)
```

```python
#Checking the dataset for Null Values

df.isnull().sum()
```

```python
#Dropping the rows with Null or NaN values

df.dropna(inplace=True)

#Dropping some columns that are irrelevant for the analysis

df.drop(['Country', 'Valid Rating', 'Net Units Sold', 'Refund Amount (INR)', 'Size', 'Color'], axis=1, inplace=True)
```

```python
#Checking for any duplicated values in the dataset

df.duplicated().value_counts()
```

```python
#Checking the new dataset

df.info()
```

```python
#Lets do a RFM (Recency-Frequency-Monetary) Analysis
#Firstly, lets create a recency dataframe with appropriate datetime formats to see customers' recent purchases in days
#The recency is calculated relative to the latest date in the dataset

reference_date = df['Order Date'].max() + pd.Timedelta(days=1)
recency = df.groupby('Customer ID')['Order Date'].max()
recency = (reference_date - recency).dt.days

#Now lets make a frequency dataframe to see how many orders did each customers make

frequency = df.groupby('Customer ID')['Order ID'].count()

#Now lets make a monetary dataframe to see how much revenue did each customers generate

monetary = df.groupby('Customer ID')['Net Revenue (INR)'].sum()

#Lets combine all these dataframes into a single dataframe

rfm = pd.DataFrame({
    'Recency': recency,
    'Frequency': frequency,
    'Monetary': monetary}).reset_index()

#Using ChatGPT to generate a RFM ranking/scoring system for each of the columns' scores

rfm['R_Score'] = pd.qcut(rfm['Recency'], 5, labels=[5, 4, 3, 2, 1])
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5])
rfm['M_Score'] = pd.qcut(rfm['Monetary'], 5, labels=[1, 2, 3, 4, 5])

#Changing the strings into integers

rfm['R_Score'] = rfm['R_Score'].astype(int)
rfm['F_Score'] = rfm['F_Score'].astype(int)
rfm['M_Score'] = rfm['M_Score'].astype(int)

#Again using ChatGPT's help to mark/rank customers into different segments based on the scores and adding them into a new Column

rfm.loc[(rfm['R_Score'] >= 3) & (rfm['F_Score'] >= 4), 'Segment'] = 'Loyal Customers'
rfm.loc[(rfm['R_Score'] >= 4) & (rfm['F_Score'] <= 2), 'Segment'] = 'Potential Loyalists'
rfm.loc[(rfm['R_Score'] <= 2) & (rfm['F_Score'] >= 3), 'Segment'] = 'At Risk'
rfm.loc[(rfm['R_Score'] <= 2) & (rfm['F_Score'] <= 2) & (rfm['M_Score'] <= 2), 'Segment'] = 'Lost Customers'
rfm.loc[(rfm['R_Score'] == 5) & (rfm['F_Score'] == 5) & (rfm['M_Score'] == 5), 'Segment'] = 'Champions'

#Filling the blank values with another category of customer segment which would be 'Other Customers'

rfm['Segment'] = rfm['Segment'].fillna('Other Customers')

#Creating a new dataframe with simply just the counts of the different customer segments

segment_counts = rfm['Segment'].value_counts().reset_index()
segment_counts.columns = ['Customer Segment', 'Number of Customers']

segment_counts
```

```python
#Lets make a Customer Dataset to see their revenue generation, total orders, average purchase value, and lifespan
#Lets start by making a dataframe of customer IDs and their respective total revenue generated as well total orders

customer_data = df.groupby('Customer ID').agg(
    Total_Revenue=('Net Revenue (INR)', 'sum'),
    Total_Orders=('Order ID', 'count')).reset_index()

#Lets create another column for average purchase value for each customer

customer_data['Average_Purchase_Value'] = customer_data['Total_Revenue']/customer_data['Total_Orders']

#Lets calculate the customer lifespan by finding their first and last purchase dates and seeing the difference in dates

first_purchase=df.groupby('Customer ID')['Order Date'].min()
last_purchase=df.groupby('Customer ID')['Order Date'].max()
customer_lifespan = (last_purchase-first_purchase).dt.days

#Mapping the new dataframe into the customer data dataframe in a new column

customer_data['Customer_Lifespan'] = customer_data['Customer ID'].map(customer_lifespan)

customer_data
```

```python
#Charting the data in a scatterplot with an Ordinary Least Squares trendline

fig16 = px.scatter(
    customer_data,
    x='Customer_Lifespan',
    y='Total_Revenue',
    trendline='ols',
    color='Total_Orders',
    size='Total_Orders',
    color_continuous_scale='Viridis',
    template='plotly_dark'
)

fig16.update_layout(
    title={'text':'<b>Customer Lifespan vs Historical Customer Value</b>', 'xanchor':'center', 'x':0.5, 'font':dict(size=20)},
    xaxis_title='Customer Lifespan (Days)',
    yaxis_title='Historical Revenue (INR)',
    coloraxis_colorbar=dict(title='Total Orders')
)

fig16.show()
```

```python
#Lets analyse the linear regression data of the customer data scatterplot chart
#The px.get_trendline_results() syntax uses statsmodel.api as its engine which makes plotly so versatile

results = px.get_trendline_results(fig16)

#Index locating the first row in the dataframe for extracting the information on the OLS Regression of the chart

regression_data = results.px_fit_results.iloc[0]

#Lets display a summarized version of the findings

print(regression_data.summary())
```

```python
#The data shown above is an exploratory association between the variables, not proof of causation
#Lets display the individual values of the Linear Regression model

print("Intercept:", regression_data.params[0])

print("Slope:", regression_data.params[1])

print("R-squared:", regression_data.rsquared)

print("Adjusted R-squared:", regression_data.rsquared_adj)

print("P-values:", regression_data.pvalues)

print("Confidence Intervals:", regression_data.conf_int())
```

```python
#Lets find the Residual Values to understand the variance in what we predicted and what actually was in regards to revenue

customer_data['Predicted_Revenue'] = (
    regression_data.params[0] +
    regression_data.params[1] * customer_data['Customer_Lifespan']
)

customer_data['Residuals'] = (
    customer_data['Total_Revenue'] -
    customer_data['Predicted_Revenue']
)

new_data = customer_data[[
    'Customer_Lifespan',
    'Total_Revenue',
    'Predicted_Revenue',
    'Residuals'
]]

print(new_data.head())
```
